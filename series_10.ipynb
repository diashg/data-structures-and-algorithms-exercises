## Exercice 1
Écrivez une file de priorité (priority queue) à l'aide d'une structure de donnée "monceau" (heap), appelée `BinaryHeap`. La classe doit implémenter l'interface `PriorityQueue`.

La solution doit également comporter des classes qui modélisent les éléments à insérer dans la priority queue. À cette fin, écrivez la class `KeyValuePair` qui implémentes l'interface `Comparable`, en utilisant un nombre entier comme clé et une chaîne de caractères comme valeur.

```python
# Une interface qui indique un contrat pour des instances qui peuvent être comparables.
class Comparable:
    # Vérifie si cet objet est plus petit qu'un autre objet o
    # Retourne True si l'objet courant est plus petit que o
    # Retourne False si l'objet courant n'est pas plus petit que o
    def less_than(self, rhs) -> bool:
        raise NotImplementedError

    # Compare cet objet avec un autre objet o
    # Retourne un entier négatif si cet objet est plus petit que o
    # Retourne un entier positif si cet objet est plus grand que o
    # Retourne un 0 si cet objet est égal à o
    def compares(self, rhs) -> int:
        raise NotImplementedError

# An interface that specifies the primitives of a priority queue.
class PriorityQueue:
    # Ajoute un objet Comparable (x) dans le heap
    def insert( self, x ): # x is of type Comparable
        raise NotImplementedError

    # Supprime et retourne l'entrée la plus petite
    def delete_min(self) -> Comparable: # throws UnderflowException
        raise NotImplementedError

    # Retourne le plus petit élément sans le supprimer
    def find_min(self) -> Comparable: # throws UnderflowException
        raise NotImplementedError

    # Supprimer toutes les entrées du heap
    def make_empty(self):
        raise NotImplementedError

    # Détermine si la collection est vide
    def is_empty(self) -> bool:
        raise NotImplementedError

    # Retourne le nombre d'éléments stockés dans la collection
    def size(self) -> int:
        raise NotImplementedError

# L'exception est levée lors d'accès aux éléments d'une collection vide
class UnderflowException(Exception):
    pass
```

```python
# L'implémentation à écrire pour l'élément inséré dans le monceau.
class KeyValuePair(Comparable):
    
    def __init__(self, key, value):
        self.key = key
        self.value = value

    def get_key(self):
        return self.key

    def less_than(self, rhs) -> bool:
      if self.key != rhs.key:
        return self.key < rhs.key
      return self.value < rhs.value

    def compares(self, rhs) -> int:
      if self.key != rhs.key:
        return self.key - rhs.key
      if self.value < rhs.value:
        return -1
      elif self.value > rhs.value:
        return 1
      return 0

    def __str__(self):
      return f"({self.key}, {self.value})"

    def __repr__(self):
      return self.__str__()


class BinaryHeap(PriorityQueue):

    def __init__(self):
        self.heap = []

    def is_empty(self) -> bool:
        return len(self.heap) == 0

    def size(self) -> int:
        return len(self.heap)

    def make_empty(self):
        self.heap = []

    def insert(self, x):
        self.heap.append(x)
        self.bubble_up(len(self.heap) - 1)

    def delete_min(self):
        if self.is_empty():
            raise UnderflowException
        min_value = self.heap[0]
        self.heap[0] = self.heap[-1]
        self.heap.pop()
        if len(self.heap) > 1:
            self.bubble_down(0)
        return min_value

    def find_min(self) -> Comparable:
      if self.is_empty():
          raise UnderflowException
      return self.heap[0]

    def bubble_up(self, index: int):
      while index > 0:
          parent = (index - 1) // 2
          # si l'index du parent est déjà <= enfant : on est bon, on s'arrête
          if not self.heap[index].less_than(self.heap[parent]):
              break
          # sinon on échange parent et enfant
          self.heap[index], self.heap[parent] = self.heap[parent], self.heap[index]
          index = parent   # on continue plus haut

    def bubble_down(self, index: int):
        n = len(self.heap)

        while True:
            left = 2 * index + 1
            right = 2 * index + 2
            smallest = index

            # on cherche l'enfant avec la plus petite clé
            if left < n and self.heap[left].less_than(self.heap[smallest]):
                smallest = left
            if right < n and self.heap[right].less_than(self.heap[smallest]):
                smallest = right

            # si aucun enfant plus petit : invariant de tas respecté
            if smallest == index:
                break

            # sinon on échange avec l'enfant le plus petit
            self.heap[index], self.heap[smallest] = self.heap[smallest], self.heap[index]
            index = smallest

```

```python
pq = BinaryHeap()
assert pq.is_empty()
assert pq.size() == 0
kv1 = KeyValuePair(5, "a")
kv2 = KeyValuePair(3, "b")
kv3 = KeyValuePair(8, "c")
kv4 = KeyValuePair(7, "d")
kv5 = KeyValuePair(1, "e")

# Doit lever une exception de type UnderflowException.
try:
    pq.find_min()
except UnderflowException:
    pass

# Doit lever une exception de type UnderflowException.
try:
    pq.delete_min()
except UnderflowException:
    pass

pq.insert(kv1)
# LE HEAP DOIT ETRE [5, a] A CE STADE
assert not pq.is_empty()
assert pq.find_min().get_key() == 5
assert pq.size() == 1

pq.insert(kv2)
# LE HEAP DOIT ETRE [3, b][5, a] A CE STADE
assert pq.find_min().get_key() == 3
assert pq.size() == 2

pq.insert(kv3)
# LE HEAP DOIT ETRE [3, b][5, a][8, c] A CE STADE
assert pq.find_min().get_key() == 3
assert pq.size() == 3

pq.insert(kv4)
# LE HEAP DOIT ETRE [3, b][5, a][8, c][7, d] A CE STADE
assert pq.find_min().get_key() == 3
assert pq.size() == 4

pq.insert(kv5)
# LE HEAP DOIT ETRE [1 , e][3 , b][8 , c][7 , d][5 , a] A CE STADE
assert pq.find_min().get_key() == 1
assert pq.size() == 5

# Tests the primitives
list_of_key_value_pairs = []
for i in range(5):
    list_of_key_value_pairs.append(pq.delete_min())

list_of_key_value_pairs.reverse()
for i in range(len(list_of_key_value_pairs)):
    pq.insert(list_of_key_value_pairs[i])

# LE HEAP DOIT ETRE [1 , e][3 , b][8 , c][7 , d][5 , a] A CE STADE
assert pq.find_min().get_key() == 1
assert pq.size() == 5

min1 = pq.delete_min().get_key()
# LE HEAP DOIT ETRE [3 , b][5 , a][8 , c][7 , d] A CE STADE
assert pq.find_min().get_key() == 3
assert pq.size() == 4
assert min1 == 1

min2 = pq.delete_min().get_key()
# LE HEAP DOIT ETRE [5 , a][7 , d][8 , c] A CE STADE
assert min2 == 3
assert pq.find_min().get_key() == 5
assert pq.size() == 3

pq.make_empty()
assert pq.size() == 0
assert pq.is_empty()
```

### Explications

On traite le cas d'une interface priorityQueue qui organise les éléments en fonction de leurs indices et non de leurs valeurs, l'indice le plus bas (min) est prioritère (pas dans le cas premier arrivé premier servi). La structure de données traités est un BinaryHeap, conceptuellement il peut être vu comme un arbre avec des nœuds et des enfants. On a un tableau et on organise en fonction de l'indice i des éléments (clef) du tableau :
si on a un élément i
son parent sur l'arbre est à l'indice : (i − 1) // 2
l'enfant à gauche de i est d'incice : 2*i + 1
l'enfant droite est d'indice : 2*i + 2

On a KeyValuePair qui représente un élément du heap, caractérisé par sa valeur et sa clef (indice), les fonctions less_than et compares permettent de comparer le positionnement de notre élément en fonction d'un passé en paramètre. Leur fonctionnement est assez clair

BinaryHeap caractérise le fonctionnement de notre structure, le rôle et fonctionnement de
is_empty, size, et make_empty est assez clair, find_min également si on a la représentation de l'arbre en tête (on renvoie juste la racine)

on a delete_min et insert qui respectivement supprime le plus petit élément du heap (si celui est non vide) et insert qui ajoute un élément au heap. On pourrait se dire que c'est trivial et qu'on s'arrête là mais on se doit de préserver la structure de BinaryHeap après l'ajout/suppression d'un élément.

exemple : on regarde les clefs et on insère un élément qui a 4 pour clée

[5, 7, 9, 12, 15, 20, 30, 4]    (4 est trop bas) on ne respecte pas la structure

il va falloir faire remonter 4 (avant le 5)

remarque importante lors de la suppression il est nécessaire de remplacer la racine récemment supprimée par le dernière élément pour maintenir une structure de HeapBinary (sans quoi la structure de l'arbre n'est plus respectée)

exemple : si cette fois on delete_min alors il faut un remplaçant à la racine... donc on prendra l'indice qui suit.


c'est le rôle de bubble_up et bubble_down, elles préservent la structure après suppression et ajout d'éléments

bubble_up :
on boucle tant que notre élément (on regarde sa clée) n'est pas a la bonne position :
on prend la clef de notre élément (i), on compare à chaque fois avec celle du parent = (i - 1) // 2

si on est plus grand que le parent alors on à la bonne place et on arrête

sinon on échange de positionnement avec le parent et on retest/recommence avec le parent suivant (le grand parent)

bubble_down :
On boucle tant que  a encore au moins 1 enfant (on veut redeterminer la racine)

on choisit à chaque fois l'enfant le plus petit
(on rappel la structure
left = 2*i + 1
right = 2*i + 2)

on détermine l'enfant le plus petit (smallest) à l'aide de la fonction less_than
si notre clée est plus grande que smallest
on échange notre place et on recommence

sinon c'est qu'on a gagné

illustration :
[4, 5, 9, 7, 15, 20, 30, 12] on retire 4

enfants = 5, 9 → minimum = 5
12 > 5 → échange → descend à la place 1

Maintenant index = 1
enfants = 7, 15 → minimum = 7
12 > 7 → échange

Maintenant index = 3
plus d’enfants → stop

[12, 5, 9, 7, 15, 20, 30]

Utilisez votre implémentation de la structure de données priority queue pour trier efficacement une liste d'objets (une liste de candidats à une élection donnée) en fonction des votes reçus et des noms.

La liste reçue en input doit être donnée dans un fichier texte (disponible sur Moodle), en suivant le format: Nom,Votes

| Exemple d'input   | Exemple d'output   |
| :----------------:|:------------------:|
| John, 6           | George, 1          |
| Paul, 5           | Ringo, 3           |
| George, 1         | Paul, 5            |
| Ringo, 3          | John, 6            |

Pour ouvrir un fichier en utilisant jupyter notebook, vous devez le mettre au même niveau que ce fichier ipynb

```python
def create_priority_queue() -> PriorityQueue:
    pq = BinaryHeap()

    with open("s10ex2.txt", "r") as f:
      for line in f:
        name, votes= line.strip().split(",")
        name=name.strip()
        votes = int(votes.strip())
        pair = KeyValuePair(votes, name)
        pq.insert(pair)
    return pq # Retourne une instance de BinaryHeap

def create_final_list(pq):
    result=[]
    while not pq.is_empty():
      result.append(pq.delete_min())
    return result # Retourne une liste ordonnée
```
