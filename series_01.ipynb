## Exercice 1
Le PGCD (plus grand commun diviseur) est le plus grand nombre entier qui divise simultanément deux autres nombres entiers.

Implémentez l'algorithme d'Euclide permettant de calculer le PGCD de deux nombres entiers. Vous trouverez plus d'informations concernant l'algorithme d'Euclide en cliquant sur ce [lien](https://en.wikipedia.org/wiki/Greatest_common_divisor#Euclid's_algorithm).

```python
def gcd(n: int, m: int):
    result = 0
    while m != 0:
        n, m = m, n%m
        result = n
        print(n, m)
    return result
print(gcd(24,32))
```

```python
assert gcd(9,6) == 3
assert gcd(24,32) == 8
assert gcd(18,18) == 18
assert not gcd(10,15) == 10
assert not gcd(12,9) == 4
assert not gcd(14,14) == 34
```

### Explications

Le code ici marche tel quel, pas besoin de le compléter.

Ici, tant que m est différent de 0, on remplace n par m et m par n modulo m.
Quand m atteindra 0, n sera le pgdc.

La fonction prend 2 chiffres et tant que m n'est pas égal a zero la boucle ne fini pas. Le resultat "n" est stocké dans la variable result et est le plus grand diviseur commun.

## Exercice 2
Implémentez une manière de calculer $x^n$ en utilisant la méthode de dichotomie.

```python
def powdi(x: int, n: int):
    if n == 0:
        return 1
    if n == 1:
        return x
    if n % 2 == 0:
        return powdi(x, n/2) * powdi(x, n/2)
    if n % 2 != 0:
        return x * powdi(x, n//2) * powdi(x, n//2)
    return None
```

```python
assert powdi(2,3) == 8
assert powdi(4,2) == 16
assert powdi(2,2) == 4
assert powdi(4,0) == 1
assert powdi(2,1) == 2
assert not powdi(5,2) == 10
assert not powdi(3,7) == 10
assert not powdi(3,3) == 10
```

### Explications

Code déjà complet ici également.

Si n = 0, on renvoie 1.

Si n = 1, on renvoie x.

Si n est pair, on calcule récurssivement avec x puissance n/2 multiplié avec lui même.

Si n est impair, la division entière (n//2) va supprimer le reste de 1, on va ensuite récurssivement avec x puissance n/2 multiplié avec lui même Ucomme précédemment, à la différence qu'ici, on multiplie ce résultat par x pour compenser la perte des restes supprimés.

Le code est récursif et se rappelle si le nombre est plus grand que 1. Si le nombre est pair alors la fonction refait une itération powdi qui regarde si n>1 sinon elle continue et, ce faisant, donne le nombre de fois que x sera au carré. Meme chose si le nombre est impair mais on doit ajouter *x.

## Exercice 3
La suite de Fibonacci est une suite de nombres entiers dans laquelle chaque nombre $f_{n+2}$ correspond à la somme des deux nombres qui le précèdent, $f_{n+1}+f_{n}$.

Implémentez l'algorithme de Fibonacci en utilisant la multiplication matricielle.

```python
import numpy as np

def fibo(n: int):
    f0f1 = np.array([[0],[1]])
    A = np.array([[0,1],
                  [1,1]])
    Apower_n = np.linalg.matrix_power(A, n) # utiliser powdi est preferable
    print(Apower_n)
    return (Apower_n @ f0f1)[0]
```

```python
fibo(8)
```

```python
def powdi(x: int, n: int):
    if n == 0:
        return np.identity(2)
    if n == 1:
        return x
    if n % 2 == 0:
        return powdi(x, n/2) @ powdi(x, n/2)
    if n % 2 != 0:
        return x @ powdi(x, n//2) @ powdi(x, n//2)
    return None
```

```python
powdi(np.array([[0,1],[1,1]]), 8)
```

```python
assert fibo(8) == 21
assert fibo(10) == 55
assert fibo(0) == 0
assert fibo(1) == 1
assert not fibo(5) == 10
```

## Exercice 4
Implémentez et testez les 3 versions de l'algorithme calculant la sous-suite de somme maximale, c'est-à-dire:
* 3 boucles imbriquées
* 2 boucles imbriquées
* une seule boucle (Kadane)

```python
def maxSub3(m):
    maxSum = 0
    currentSum = 0
    for i in range(len(m)):
        for j in range(i, len(m)):
            currentSum = 0
            for k in range(i, j+1):
                currentSum += m[k]
            if currentSum > maxSum:
                maxSum = currentSum
    if currentSum < 0:
        currentSum = 0
    return maxSum

def maxSub2(m):
    maxSum = 0
    for i in range(len(m)):
        currentSum = 0
        for j in range(i, len(m)):
            currentSum += m[j]
            if currentSum > maxSum:
                maxSum = currentSum
    return maxSum

def maxSub1(m):
    maxSum = 0
    currentSum = 0
    for x in m:
        currentSum = max(x, currentSum + x)
        maxSum = max(maxSum, currentSum)
    return maxSum
```

```python
assert maxSub3([4,3,-10,2]) == 7
assert maxSub3([4,3,-10,2,8]) == 10
assert not maxSub3([4,3,-10,2,8]) == -10

assert maxSub2([4,3,-10,2]) == 7
assert maxSub2([4,3,-10,2,8]) == 10
assert not maxSub2([4,3,-10,2,8]) == -10

assert maxSub1([4,3,-10,2]) == 7
assert maxSub1([4,3,-10,2,8]) == 10
assert not maxSub1([4,3,-10,2,8]) == -10
```

### Explications

1. maxSub3 parcourt toutes les combinaisons possibles, recalcule chacune de leur somme et garde la plus grande. Cette manière de procéder peut faire penser à un brute force.

2. maxSub2 parcourt aussi toutes les combinaisons, mais au lieu de tout calculer, il réutilise les sommes déjà enregistrées.

3. maxSub1 décide à chaque élément s'il vaut mieux continuer ou recommencer à partir de cet élément, tout en gardant toujours la somme maximale rencontrée. (Kadane)



Les 3 fonctions font la même chose mais ne prennent pas les même ressources pour le faire.
