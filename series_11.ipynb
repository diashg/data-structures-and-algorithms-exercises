Implémentez une type de données abstrait `HashTable` (table de hachage) qui utilise une dispersion linéaire (linear probing strategy) pour résoudre les collisions. La classe `HashTable` et ses fonctions à implémenter sont données - libre à vous cependant d'ajouter de nouvelles méthodes aux aux classes. Vous pouvez optionnellement utiliser une dispersion quadratique.

Vous devez implémenter une classe `HashableImpl` dont les fonctions sont données. Cela correspond aux éléments qui seront stockés dans la hash table.

````python
# Représente un élément pouvant être stocké dans une instance de HashTable.
class HashableImpl:
    key = None
    is_active = True
    def __init__(self, key, is_active= True):
        self.key = key
        self.is_active = is_active  # (useless for you but keep it)

    # Calcule le hash de cette instance de HashableImpl.
    # La valeur retournée est un hash, représenté par un nombre entier.
    def hash(self, table_size):
        return self.key%table_size

    # Retourne la clé comme représentation de cette instance
    def __str__(self):
        return str(self.key)


class HashTable:
    def __init__(self):
        self.capacity = 11
        self.ht = [None] * self.capacity # Initialise la liste utilisée pour stocker les éléments de la hashtable
        self.size = 0 # Compte le nombre de clés présentes dans la table
        self.collisions = 0
        pass

    # Indique le nombre de collisions qui ont eu lieu lors de la recherche d'éléments dans la hashtable
    def number_of_collisions(self):
        return self.collisions

    # Vide la hashtable et réinitialise le compteur de collisions
    def make_empty(self):
        self.ht = [None] * self.capacity
        self.size = 0
        self.collisions = 0

    # Vérifie si la hashtable est vide ou pas.
    # Retourne True si la hashtable est vide, False sinon
    def is_empty(self):
        return self.size == 0


    # Insére un élément x HashableImpl dans la hash table.
    def insert(self, x):
        index = x.hash(self.capacity)

        while True:
          entry = self.ht[index]
          if entry is None or not entry.is_active:
            self.ht[index] = HashableImpl(x.key, True)
            self.size += 1
            return
          self.collisions += 1
          index = (index + 1) % self.capacity

    # Supprime l'élément x HashableImpl de la hash table
    # Une exception ItemNotFound est levée si l'élément ne se trouve pas dans la liste
    def remove(self, x): # Throws ItemNotFound
        index = x.hash(self.capacity)
        start = index

        while True:
          entry = self.ht[index]
          if entry is None:
            raise ItemNotFound
          if entry.is_active and entry.key == x.key:
            entry.is_active = False
            self.size -= 1
            return
          self.collisions += 1
          index = (index + 1) % self.capacity

          if index == start:
            raise ItemNotFound

    # Trouve l'élément x HashableImpl dans hash table
    # Une exception ItemNotFound est levée si l'élément ne se trouve pas dans la liste
    # Retourne l'élément HashableImpl
    def find(self,x): # Throws ItemNotFound
        index = x.hash(self.capacity)
        start = index

        while True:
          entry = self.ht[index]
          if entry is None:
            raise ItemNotFound
          if entry.is_active and entry.key == x.key:
            return entry
          self.collisions += 1
          index = (index + 1) % self.capacity

          if index == start:
            raise ItemNotFound

class Error(Exception):
    pass

class ItemNotFound(Error):
    pass
````

### Explications

Pour commencer on crée la classe HashableImpl avec une key et une valeur booléenne is_active puisqu'on a besoin de savoir si une valeur est active ou pas dans le tableau ensuit différente méthodes déjà expliquées dans le texte.

La méthode insert : index = calcule indice de départ puis boucle pour trouver où inseré le entry qui est où pointe l'index, le if vérifie qu'il n'y a pas de valeur ou pas de valeur active puis met la valeur dans la table si c'est bon sinon continuer et ajoute 1 collision et avance d'une case utilisant % self.capacity si on est à la fin de liste alors il renvoie au début.

La méthode remove comme insert, elle définie un pointeur index et le début start: La boucle continue tant que vrai, entry est le pointeur index si entry n'existe pas alors on Raise ItemNotFound si entry est actif et qu'on est au bon endroit, on lui enlève le statut actif on diminue la taille et on return sinon on continue comme dans insert mais si l'index revient à start alors on raise ItemNotFound

La méthode find même début que les 2 autres. La boucle utilise les même principe que les autres et notre intérêt est dans le if qui vérifie que entry est actif et qu'il correspond à l'elément qu'on recherche.

````python
import numpy as np
import random

table_size = 997

#On répète la même expérience pour plusieurs tailles de remplissage : 200 clés, puis 400, etc.
#Taille de la table = 997 fixe.

for n in [200, 400, 600, 800, 900]:

    #Création de la table de taille 997
    ht = HashTable()
    ht.capacity = table_size
    ht.ht = [None] * table_size
    keys = []

    # dispersion

    for _ in range(n):
        k = np.random.randint(4242) #    On génère n entiers aléatoires entre 0 et 4241.
        h = HashableImpl(k, True) # Pour chaque entier, on crée un objet HashableImple.
        ht.insert(h) #  On insère cette clé dans la table via insert.
        keys.append(h) #  On garde toutes les clés dans keys pour pouvoir les rechercher plus tard.

    # reset des collisions pour la partie "find" car sa fausse ce qu'on souhaite estimer
    ht.collisions = 0

    # On mesure les miss pour 100 recherches
    for _ in range(100):
        h = random.choice(keys)
        ht.find(h)
    #On en tire la moyenne
    alpha = n / table_size
    miss_moyen = ht.collisions / 100

    print(f"n = {n},  alpha = {alpha:.2f},  miss_moyen = {miss_moyen:.2f}")
````

Un *miss* (similaire à une collision) est, lors d'une phase de sondage, la visite d'une cellule de la table qui ne correspond pas à la clé recherchée.

Pour chaque $n$, afficher le nombre de *miss* pour 100 requêtes `find` de clés existantes (les choisir au hasard).





Quelle est la relation entre le facteur de charge (load factor) de la table de hachage et le nombre de miss ?

Le facteur de charge mesure à quel point la table est remplie (=nombre d'éléments dans la table / sa capacité)
Quand alpha est petit, la probabilité de collision est faible donc on a très peu de miss et on trouve très vite la clée avec le sondage


Quelle est la particularité de 997, la taille choisie pour la hash table ?

997 est un nombre premier proche de 1000, c'est un bon choix parce que ça répartit mieux les clés et évite certaines périodicités cad : Si on choisit une taille qui a beaucoup de diviseurs (comme 1000 = 2×2×2×5×5×5), alors %1000 va souvent retomber dans les mêmes restes modulo 5, en prenant 997 aucun diviseur simple ne vient “créer des groupes”, et la distribution se fait beaucoup plus uniformément.


## Exercice 2 (Prim)

Soit G, un graphe connexe, à arêtes pondérées. Un arbre couvrant de poids minimal (minimum spanning tree, MST) de G est un sous-graphe de G qui satisfait les propriétés suivantes:

* C'est un arbre, c'est-à-dire qu'il est connecté et n'a pas de cycles.
* Il est couvrant, c'est-à-dire qu'il contient tous les sommets de G.
* Il a un poids total des arêtes minimal parmi tous les arbres possibles.


Implémentez l'algorithme de Prim vu en cours pour calculer l'arbre couvrant de poids minimal (MST). L'algorithme de Prim est décrit par le pseudo-code suivant:


```txt
Choisissez un sommet arbitraire s;
Marquez s comme visité;
Repeat n-1 times {
	Soit {u, v} une arête avec le plus petit poids parmi toutes les arêtes
    où u est un sommet visité et v ne l'est pas;
	Marquez v comme visité;
	Ajoutez {u, v} dans l'arbre;
}
```

````python
# Représente un sommet.
class Vertex:
    def __init__(self, name, payload = None):
        self.name = name
        self.payload = payload
        self.edges = []

    def connect(self, other_vertex, weight):
        edge = Edge(self, other_vertex, weight)
        self.edges.append(edge)
        other_vertex.edges.append(edge)

    def __str__(self):
        return f"({self.name}: {self.payload})"

    # Peut être étendue selon vos besoins pour l'algorithme de Prim.

# Représente une arête, connectant deux sommets.
class Edge:
    def __init__(self, left_vertex, right_vertex, weight):
        self.left_vertex = left_vertex
        self.right_vertex = right_vertex
        self.weight = weight
        self.part_of_mst = False

    # Marque cette arête comme faisant partie de l'arbre couvrant de poids minimal.
    def mark_as_part_of_mst(self):
        self.part_of_mst = True

    # Ne marque plus cette arête comme faisant partie de l'arbre couvrant de poids minimal.
    def unmark_as_part_of_mst(self):
        self.part_of_mst = False

    # Indique si l'arête fait partie de l'arbre couvrant de poids minimal.
    def is_part_of_mst(self):
        return self.part_of_mst

    def __str__(self):
        return f"{self.left_vertex} <=> {self.right_vertex} (weight: {self.weight})"

    # Peut être étendue selon vos besoins pour l'algorithme de Prim.
````

````python
# Exécute l'algorithme de Prim sur une liste de sommets.
# Retourne le poids de l'arbre couvrant de poids minimal.
def prim_algorithm(vertices: list[Vertex]) -> int:
    if len(vertices) == 0:
        return 0

    start = vertices[0] #sommet arbitraire (premier de la liste ici)
    visited = [start] #liste des sommets visités
    weight = 0

    for e in range(len(vertices)-1):
      min_weight = float('inf') #poid minimal, attention ici
                                #on initialise a un nombre infini pour etre sur qu'il soit plus grand
                                #dans la condition id edge.weight < 0
      best_edge = None
      vertex_add = None

      for v in visited:       #on parcours les sommets visités
        for edge in v.edges:    #pour ensuite regarder les arretes liées

          v_unvisit = edge.right_vertex if edge.left_vertex in visited else edge.left_vertex #voisi au bout de l'arrete

          if v_unvisit not in visited:
            if edge.weight < min_weight:  #ici même logique que l'algorithme glouton vu en classe
              min_weight = edge.weight
              best_edge = edge
              vertex_add = v_unvisit

      visited.append(vertex_add) #ajout dans la listes des arretes visitées
      weight += min_weight
      best_edge.mark_as_part_of_mst()

    return weight # Retourne le poids de l'arbre couvrant de poids minimal.
````

````python
u0 = Vertex("u0")
u1 = Vertex("u1")
u2 = Vertex("u2")
u3 = Vertex("u3")
u4 = Vertex("u4")

u0.connect(u1, 4)
u0.connect(u2, 4)
u0.connect(u3, 3)
u0.connect(u4, 6)
u1.connect(u2, 3)
u1.connect(u4, 7)
u2.connect(u3, 2)
u3.connect(u4, 5)

mst_weight = prim_algorithm([u0, u1, u2, u3, u4])
assert mst_weight == 13
````
