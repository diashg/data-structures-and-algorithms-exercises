## Exercice 2
Implémentez une méthode récursive qui trouve et retourne le plus petit élément d'une liste, où la liste et sa taille sont données en paramètre.

```python
def minEl(m, s):
    if s == 1:
        return m[0]
    else:
        min_rest =  minEl(m, s-1)
    if m[s-1] < min_rest:
        return m[s-1]
    else:
        return min_rest
```

```python
assert minEl([6,5,3,9,1], 5) == 1
assert minEl([6,5,3,-9,1], 5) == -9
assert not minEl([6,5,3,-9,1], 5) == 1
```

### Explications

Nous commençons par traiter le cas où s=1 (taille de la liste). Dans ce cas, on retourne directement cette valeure comme la plus petite étant donné qu'elle est seule. Si s>1, on rappele récursivement la fonction avec s-1 pour trouver la valeure minimum des premier nombres qu'on stockera dans min_rest. Pour finir, on compare la dernière veleure de la liste avec min_rest et on va retourner la plus petite valeure.

## Exercice 3
Implémentez une méthode récursive qui cherche un élément dans une liste triée en utilisant la recherche binaire. La liste, la taille et l'élément cible sont donnés en paramètre.

```python
def findItem(m, s: int, t: int):
    if s == 0:
        return None
    half = s // 2
    if t < m[half]:
        return findItem(m[:half], half, t)
    elif t > m[half]:
        res = findItem(m[half+1:], s - half - 1, t)
        if res is None:
            return None
        else:
            return res + half + 1   # on corrige l’indice
    else:
        return half #on renvoie l'indice de m une fois t trouvé
```

```python
assert findItem([1,2,4,7,8], 5, 7) == 3
assert findItem([-12,-9,10,44,85,91], 6, -9) == 1
assert not findItem([2,4,7,9], 4, 2) == 1
```

### Explications

Explications : on regarde au milieu de la liste si la valeur t est trouvée si c'est pas le cas on coupe la liste en 2 et on réhitère le processus selon le cas ou on est < t ou que > t.

## Exercice 4
La "Fonction 91 de McCarthy" est définie comme suit:

    M(n) for integers > 0:
      if n > 100, M(n) = n - 10
      if n <= 100, M(n) = M(M(n+11))

La notation `M(M(n+11))` est un appel récursif imbriqué.

Implémentez et testez une méthode python qui retourne la nombre de McCarthy.

```python
def mcCarthy(n: int):
    if n > 100:
      return n - 10
    if n <= 100:
      return mcCarthy(mcCarthy(n+11))
    return None
```

```python
assert mcCarthy(91) == 91
assert mcCarthy(101) == 91
assert mcCarthy(102) == 92
assert mcCarthy(104) == 94
print(mcCarthy(101))
```

### Explications

Si le nombre est plus petit que 101 la finalité sera toujours 91 alors que les nombre plus grand que 100 reviennent à n-10. Le double appel récursif sert à renvoyer n dans la fonction pour déterminé par ou il passe.

## Exercice 5 - (<font color='#db60cf'>Bonus</font>) Complexité algorithmique de Fibonacci récursif
On définit l'*algorithme de Fibonacci récursif* comme suit :

```python
def fibonacci_recursive(n):
    if n <= 1:
        return n
    else:
        result1 = fibonacci_recursive(n-1,)
        result2 = fibonacci_recursive(n-2,)
        return result1 + result2  # Summation counts for 1 in the complexity
```
