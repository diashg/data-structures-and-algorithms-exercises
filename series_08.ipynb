## Exercice 1
Le problème de la ligne d'horizon. Un immeuble Bi est représenté par un triplet (Li, Hi, Ri) où Li et Ri désignent la coordonées x de la partie gauche et droite de l'immeuble, et Hi désigne la hauteur de l'immeuble. Une ligne d'horizon d'un ensemble de n immeuble est une liste des coordonnées x et des hauteurs qui les relient, classées dans l'ordre de gauche à droite.

Exemple: La ligne d'horizon des immeubles

<center>{(3, 13, 9),(1, 11, 5),(12, 7, 16),(14, 3, 25),(19, 18, 22),(2, 6, 7),(23, 13, 29),(23, 4, 28)}</center>

<center>est</center>

<center>{1, 11, 3, 13, 9, 0, 12, 7, 16, 3, 19, 18, 22, 3, 23, 13, 29, 0}</center>

<center>où les points référencent alternativement l'axe des abscissses et des ordonnées</center>

<center>(notez que les coordonnées x dans un horizon sont triées).</center>

![skylines.png](./skylines.png)

Soit la taille d'une ligne d'horizon le nombre total d'éléments (coordonnées et hauteurs) de sa liste. Implémentez un algorithme pour combiner une ligne d'horizon A de taille n1 et une ligne d'horizon B de taille n2 en une ligne d'horizon S de taille O(n1 + n2). Votre algorithme doit s'exécuter en un temps O(n1 + n2).

```python
def combine_skyline(a,b):
    lA = len(a) #longueur des listes
    lB = len(b)

    s = []  #liste vide qui contiendra la liste finale

    iA = 0  #pointeur de a
    iB = 0  #pointeur de b

    hA = 0  #hauteur
    hB = 0

    maxH = 0 #hauteur maximale de la ligne (précédemment)
              #maxH nous permet d'éviter d'ajouter des points inutiles si la hauteur du buildinf précédent
              #est égale à la hauteur du building actuel

    while iA < lA and iB < lB:
        xA = a[iA]    #valeure actuelle de a sur l'axe x (en fonction de son pointeur)
        xB = b[iB]    #valeure actuelle de b sur l'axe x (en fonction de son pointeur)

        if xA < xB:   #Si a vien avant b:
            n = xA    #valeure de l'axe x qu'on ajoutera à la liste
            hA = a[iA+1]
            iA += 2    #pour pointer sur la prochaine valeure de l'axe x pour a

        elif xA > xB:  #meme logique
            n = xB
            hB = b[iB+1]
            iB += 2

        else:  # xA = xB
            n = xA
            hA = a[iA+1]
            hB = b[iB+1]
            iA += 2
            iB += 2

        current_maxH = max(hA,hB)
        if current_maxH != maxH:
            s.append(n)
            s.append(current_maxH)
            maxH = current_maxH

    #si une liste est plus longue que l'autre:

    while iA < lA:
        n = a[iA]
        current_maxH = a[iA+1]
        if current_maxH != maxH:
            s.append(n)
            s.append(current_maxH)
            maxH = current_maxH
        iA += 2

    while iB < lB:
        n = b[iB]
        current_maxH = b[iB+1]
        if current_maxH != maxH:
            s.append(n)
            s.append(current_maxH)
            maxH = current_maxH
        iB += 2

    if s[-1] != 0:
      s.append(s[-2])
      s.append(0)

    return s
```

```python
a = [1,8,6,10,8,0]
b = [2,4,5,15,11,0]
s = combine_skyline(a,b)
# s = sum(s, [])
print("Skyline A:")
print(a)
print("Skyline B:")
print(b)
print("Skyline S (must be [1, 8, 5, 15, 11, 0]):")
print(s)
assert s == [1,8,5,15,11,0], "The merge of skyline A and skyline B failed"

print()

a = [2,7,7,9,10,5,14,12,15,0]
b = [3,6,8,12,9,3,13,15,16,0]
s = combine_skyline(a,b)
# s = sum(s, [])
print("Skyline A:")
print(a)
print("Skyline B:")
print(b)
print("Skyline S (must be [2, 7, 7, 9, 8, 12, 9, 9, 10, 5, 13, 15, 16, 0]):")
print(s)
assert s == [2,7,7,9,8,12,9,9,10,5,13,15,16,0]
```

### Explications

On commence par créer plusieurs variables: lA et lB pour la longueur des listes, iA et iB comme pointeur de liste(vaut initialement 0), hA et hB pour utiliser la hauteur pour un point donné dans la liste(aussi initié à 0) et maxH qui contendra la hauteure maximale du dernier point traité dans le code(celui ajouté à la liste s). On crée également une liste vide qui va contenir le résultat final.

Ensuite, on crée une boucle while tant que le pointeur de a et b est respectivement inférieur à la longueur de a et b. Celle boucle permet que losqu'un des pointeur est égal à la longueur de sa liste respective, la boucle s'arrête car la liste aura été parcourrur entièrement.

Dans la boucle, on commence par trouver xA et xB (valeure sur l'axe x) en fonction du pointeur. Ensuite nous comparons xA et xB afin de voir lequel vient en premier(le x le plus petit). L'élément qui apparaitra en premier verra sa valeure x stocké dans la variable n. On met ensuite à jour la hauteure active(hA ou hB) correspondant au x qu'on a choisi.(par exemple si on a choisi xA, on met à jour hA = a[iA + 1], puis on avance le pointeur de la liste qu'on vient de lire). Après avoir choisi n et mis à jour la hauteur, on calcule le current_maxH(nouvelle hauteure dominante de n). Attention ici, à la fin de la boucle il y a un if qui permet de comparer la valeure de la hauteure dominante actuelle (current_maxH) à celle de la hauteure du dernier point ajouté(maxH). On ajoute ainsi la valeure du x et current_maxH uniquement si les deux hauteures sonf différentes, si elles sont égales, pas besoin d'ajouter. (care la ligne d'horizon ne change pas)

A la fin du code, on s'occupe de gérer le cas ou une des listes en entrée est plus longue que l'autre. Par exemple, si len(a) = 10 et len(b) = 12, lorsque iA et iB seront égal à 10, la première partie du code s'arrêtera et il restera à traiter le cas "while iB < lB" (exemple pour illustrer, mais ça pourrait être le cas inverse). Dans cette situation là, ce bout de code permet d'ajouter la fin de la liste en fonction de la fin de la liste plus longue(de la même manière que précédemment).

Pour finir, on retourne la liste s.


### Exercice 1.1
Implémentez un algorithme de "diviser pour régner" O(n log n) pour trouver la ligne d'horizon de n bâtiments.


```python
# Le paramètre "buildings" est une liste composée de triplets (Li, Hi, Ri) comme indiqué dans l'exercice 1.
# La valeur retournée est une liste, formant la ligne d'horizon.
def find_skyline(buildings):
    if not buildings:
      return []

    if len(buildings) ==1: #cas de base
      (L, H, R) = buildings[0]
      return [L, H, R, 0]

    mid = len(buildings) // 2
    left = find_skyline(buildings[:mid])
    right = find_skyline(buildings[mid:])

    return combine_skyline(left,right)
```

```python
skyline = find_skyline([[3, 13, 9],[1, 11, 5],[12, 7, 16],[14, 3, 25],[19, 18, 22],[2, 6, 7],[23, 13, 29],[23, 4, 28]])
# skyline = sum(skyline, [])
assert skyline == [1, 11, 3, 13, 9, 0, 12, 7, 16, 3, 19, 18, 22, 3, 23, 13, 29, 0]
```

## Exercice 2
Le tri par fusion (merge sort) est l'algorithme classique de type "diviser pour régner". Implémentez et analysez-le en essayant de grands inputs. Assurez-vous d'essayer des inputs à la fois distribués de manière aléatoire et triées par ordre décroissant.

```python
def merge_sort(s): #phase de division
    if len(s) <= 1 :
      return s
    mid = len(s) // 2 #division entière on veut le milieu de la liste
    #on divise le tableau en deux de manière récursive jusqu'à obtenir une liste
    #avec un singleton
    left_s = merge_sort(s[:mid])
    right_s = merge_sort(s[mid:])

    return merge(left_s, right_s) #on fait appel la phase de fusion

def merge(left, right) :
  sorted_s = []
  i = j = 0
  while i < len(left) and j < len(right) :
    if left[i] <= right[j]:
      sorted_s.append(left[i])
      i += 1
    else :
      sorted_s.append(right[j])
      j += 1

  sorted_s.extend(left[i:])
  sorted_s.extend(right[j:])
  return sorted_s
```

```python
import random
random_list = []
for i in range(0,20):
    n = random.randint(1,5000)
    random_list.append(n)

print("Initial list (random order): ")
print(random_list)
print("My sorted list")
my_sorted_list = merge_sort(random_list.copy())
print(my_sorted_list)
print("Python sorted list")
python_sorted_list = random_list.copy()
python_sorted_list.sort()
print(python_sorted_list)
assert my_sorted_list == python_sorted_list, "The list is not correctly sorted"

print()
random_list = []
for i in range(0,25):
    n = random.randint(1,5000)
    random_list.append(n)

print("Initial list (descending order): ")
random_list.sort()
random_list.reverse()
print(random_list)
print("My sorted list")
my_sorted_list = merge_sort(random_list.copy())
print(my_sorted_list)
print("Python sorted list")
python_sorted_list = random_list.copy()
python_sorted_list.sort()
print(python_sorted_list)
assert my_sorted_list == python_sorted_list, "The list is not correctly sorted"
```

### Explications

Comme il est mentionné dans l'énoncé le "**merge sort**" est un algorithme qui "divise pour mieux régner" alors on va segmenter notre algo en deux phases distinctes : la phase ou l'on "divise" notre liste de nombres, puis la phase ou l'on refussionne (merge) notre tableau une fois celui-ci divisé...

phase de division :
Dans notre fonction merge_sort on divise par réccurence notre liste par 2 : c'est à dire qu'on la segmente au milieu et on obtient left et right on réitère par récursion jusqu'à obtenir des listes de taille 1, une fois ce procédé fait on remonte par récurrence en fusionnant les listes dans le bon ordre : on refusion à gauche et à droite à chaque fois dans l'ordre croissant avec la fonction **merge**, la récursion réitère le processus pour les listes de taille 1 puis 2 ... jusqu'à la fusion finale qui nous donne le résultat escompté  

                    [8, 3, 5, 2]
                    /          \
           [8, 3]                 [5, 2]
           /    \                /      \
         [8]    [3]            [5]      [2]
          ↑        ↑            ↑         ↑
    fusion →   [3, 8]      fusion →     [2, 5]
                    \          /
                    fusion finale
                    [2, 3, 5, 8]

La fusion se fait quand on revient de l’appel récursif. (cad on a finit de diviser)
Chaque appel attend les résultats de ses sous-appels.
La récursion descend, le merge remonte.

phase de fusion :
On lui passe toujours une partie gauche qui est une liste de taile n/2 et partie droite qui est une liste de taille n/2 également. Elle fusionne les deux listes pour avoir une liste de taille n.
**Comment fonctionne le tri **:
On regarde l’élément courant de gauche et celui de droite (indices i et j), on prend le plus petit, et on avance dans cette liste.
jusqu'à ce que l'une des listes soit vide puis on ajoute le reste de l'autre liste (le reste est déterminé par rapport à l'indice i ou j), le reste sera automatiquement trié:

Exemple:

left :   [ 2,     7,     12 ]
right:   [ 3,  4,  10,  11 ]

i = 0, j = 0
→ [2]                 (2 < 3)        i = 1, j = 0

i = 1, j = 0
→ [2, 3]              (3 < 7)        i = 1, j = 1

i = 1, j = 1
→ [2, 3, 4]           (4 < 7)        i = 1, j = 2

i = 1, j = 2
→ [2, 3, 4, 7]        (7 < 10)       i = 2, j = 2

i = 2, j = 2
→ [2, 3, 4, 7, 10]    (10 < 12)      i = 2, j = 3

i = 2, j = 3
→ [2, 3, 4, 7, 10, 11] (11 < 12)     i = 2, j = 4  (right épuisé)

→ ajouter reste de left → 12

= [2, 3, 4, 7, 10, 11, 12]
