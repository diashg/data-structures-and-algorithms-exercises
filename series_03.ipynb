## Exercice 0 - Multiplication de matrices
Quelle est la condition sur la dimension de deux matrices pour effectuer leur multiplication ?

Il faut que le nombre de ligne de la 1ère matrcie soit égal au nombre de colonne de la 2 eme matrice.

Implémenter le produit de matrices en utilisant des `list`. L'algorithme doit retourner `False` si les dimensions des matrices en argument ne sont pas compatible avec la multiplication (cela inclut de vérifier si l'input est une matrice !)

```python
def matrix_multiply(A, B):
    for row in A:
      if len(row) != len(A[0]):# verifier que la matrice est de la meme longueur chaque ligne
        return False
    for row in B:
      if len(row) != len(B[0]):
        return False
    if len(A[0])!=len(B):
        return False
    result = [[0 for g in range(len(B[0]))] for h in range(len(A))] #création de la matrice
    for i in range(len(A)):
      for j in range(len(B[0])):
        for k in range(len(B)):
          result[i][j]+=A[i][k]*B[k][j] #valeur dans position
    return result
```

```python
matrix_A = [[1, 2, 3],
            [4, 5, 6]]
matrix_B = [[7, 8],
            [9, 10],
            [11, 12]]
result_matrix = matrix_multiply(matrix_A, matrix_B)
assert result_matrix == [[58, 64], [139, 154]]

matrix_C = [[1, 2, 4], [3, 4]]
matrix_D = [[5, 6, 7], [8, 9, 10]]
result_matrix = matrix_multiply(matrix_C, matrix_D)
assert (result_matrix is False)

matrix_E = [[1, 2, 4], [3, 4, 6]]
matrix_F = [[5, 6], [8, 9]]
result_matrix = matrix_multiply(matrix_E, matrix_F)
assert result_matrix is False

matrix_G = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
identity_matrix = [[1, 0, 0], [0, 1, 0], [0, 0, 1]]
result_matrix = matrix_multiply(matrix_G, identity_matrix)
assert result_matrix == matrix_G
```

Quelle est la complexité de la multiplication de matrices ?

La comlexité est de O(n * m * p) | n = nombre de ligne de A |  m=nombre de colonne de A/nombre de ligne de B |  p= nombre de colonne de B

## Exercice 1
Ecrivez une méthode récursive qui détermine si une chaîne de caractères (un tableau de caractères), est un palindrome, où le tableau et sa taille sont donnés en paramètre. La signature de la méthode à implémenter **doit** être:

    //retourner 1 si a[] est un palindrome, sinon 0
    int isPalindrome(char a[], int n)

```python
def isPalindrome(a, n: int):
    if n<=1:
        return 1

    if a[0] != a[-1]:
        return 0
    else:
        a=a[1:-1]
        n=n-2
        return isPalindrome(a, n)
```

```python
assert isPalindrome(["a","b","b","a"], 4) == True
assert isPalindrome(["k","a","y","a","k"], 5) == True
assert isPalindrome(["k","i","y","a","k"], 5) == False
assert not isPalindrome(["a","b","a","b"], 4) == True
```

### Explications

On va d'abbord traiter le cas où ou n <= 1.
Pour ce qui est des listes plus longues, on compare simplement si le premier élément de la liste est différent du dernier. Si oui, ce n'est pas un palindrome. Si ils sont égaux, on va modifier la liste a pour enlever le premier et dernier élément de cette liste, et soustraire 2 à n. Aprèc cela, nous allons pouvoir utiliser récurvisement notre fonction avec les nouvelles valeures. Ainsi, notre code va comparer tout les éléments de la liste et retourner 1 si c'est un palindrome.

### Exercice 1.1
Quelle est la complexité de l'algorithme? Expliquez.

La complexité de cet algorithme est de  O(n²) car a chaque appel récursif, la fonction va appeler une nouvelle sous-liste(a=a[1:-1]).
Etant donné que l'on enlève 2 éléments de la liste à chaque appel récursif, plusieurs opérations auront lieu de manière décroissante: O(n)+O(n-2)+O(n-4)+... jusqu'à O(1).

Comme il y aura plusieurs appels récursifs mais que chaque appel demandera moin de ressource que le précédent, on peut estimer la complexité à O(n²).

## Exercise 2
Implémentez une solution récursive du jeu des Tours de Hanoï. Le jeu consiste à déplacer des disques de différents diamètres d'une tour de départ à une tour d'arrivée, **_en passant par une tour intermédiaire_**. Dans cette version du jeu, il n'y a que 3 tours : la tour de gauche, celle du milieu et celle de droite. La solution correcte doit déplacer tous les disques de leur tour de départ vers la tour de destination en effectuant le nombre minimum d'étapes et en respectant les règles suivantes :

* il n'est possible de déplacer qu'un seul disque à la foi
* il est possible de déplacer un disque seulement au-dessus d'un disque plus grand ou dans une tour vide

Dans la phase initiale, il y a **N** disques dans la tour de départ, et les deux autres tours sont vides. Chaque disque comporte un disque plus grand en dessous de lui, à l'exception du disque le plus bas, et un disque plus petit au-dessus de lui, à l'exception du disque le plus haut.


Le programme principal doit prendre **N** comme paramètre d'entrée. L'output du programme est la séquence de pas effectués pour déplacer tous les disques dans la tour d'arrivée. L'exercice ne nécessite pas d'implémenter une interface graphique pour visualiser les mouvements des disques.

Exemple d'output for N=3:

[step 1] Move disk 1 from A to C

[step 2] Move disk 2 from A to B

[step 3] Move disk 1 from C to B

[step 4] Move disk 3 from A to C

[step 5] Move disk 1 from B to A

[step 6] Move disk 2 from B to C

[step 7] Move disk 1 from A to C

```python
counter = 0
def hanoi(n: int, startTower: int, goalTower: int, auxTower: int):
    global counter

    if n == 1:
      counter+=1
      print(f"[step {counter}] Move disk {n} from {startTower} to {goalTower}")
    else:
      hanoi(n-1,startTower,auxTower,goalTower)
      counter+=1
      print(f"[step {counter}] Move disk {n} from {startTower} to {goalTower}")
      hanoi(n-1,auxTower,goalTower,startTower)
```

```python
hanoi(2, 'A', 'C', 'B')
print()
counter = 0
hanoi(3, 'A', 'C', 'B')
```

### Explications

Ce code va permettre de déplacer n disques de la tour de base(startTower) vers la tour d'arrivée(goalTower) en s'aidant la tour intermédiaire pour y parvenir(auxTower). Nous avons aussi un compteur(counter) qui sert à numéroter les étapes.

Dans le cas où n=1, il ne nous reste plus qu'un seul disque à déplacer on ajoute un au compteur et on affiche le nouvement du dernier disque à déplacer (car n=1)

Dans le cas ou n>1, cela veut dire qu'il nous reste plusieurs disques à déplacer. Nous allons donc rappeler la fonction récursivement avec n-1, en déplaçant les disques suppérieurs de la tour de départ vers la tour auxiliaire(hanoi(n-1,startTower,auxTower,goalTower)).

Ensuite, on augemente le compteur de 1 et on déplace le plus grand disque de la tour de départ(disque n) vers la tour d'arrivée en l'affichant la fonction print.

Pour finir, on déplace les n-1 disques déplacés précédemment à la tour auxiliaire vers la tour d'arrivée(hanoi(n-1,auxTower,goalTower,startTower))

### Exercice 2.2
Quel est le rapport entre le nombre total de mouvements de disques et N ? Justifiez votre réponse.

Pour déplacer N disques, nous devrons d'abbord déplacer les N-1 disques du dessus(f(N-1). Ensuite nous déplaçons le plus grand disque(+1) avant de replacer les N-1 disques au dessus de celui-ci(f(N-1)).

Nous avons donc f(N) = f(N-1) + 1 + f(N-1) = 2*f(N-1) + 1
avec f(1)=1 comme hyphotèse on trouve que fn= 2(^n)-1  

Donc pour déplacer N disques, nous aurons 2(^n)-1 mouvements.

# Exercice 3 - Puits de fleurs magique

Il y a un puits magique profond dans une forêt qui a quelques nénuphars sur ses eaux. Vous avez un grand panier vide et quelques pièces de monnaie, et vous vous tenez à côté du puits. Vous avez plus de pièces que de nénuphars dans le puits. Le puits a pris note du fait que votre panier est vide. Si vous lancez une pièce dans le puits, le puits lancera un nénuphar dans votre panier. Si vous lancez quatre pièces à la fois dans le puits, le puits prendra note du nombre de nénuphars qu'il a lancés dans votre panier jusqu'à présent. Si vous lancez deux pièces à la fois dans le puits, le puits lancera autant de nénuphars dans votre panier qu'il avait pris note la dernière fois. Si vous lancez une pièce, ou deux pièces à la fois, dans le puits, et qu'il ne reste pas assez de nénuphars dans le puits, le puits ne lancera aucun nénuphar.
Étant donné $L$ le nombre de nénuphars dans le puits au début, retournez le nombre minimum de pièces que vous devrez lancer dans le puits pour qu'il lance tous ses nénuphars dans votre panier.

```python
def min_well_coin(remaining_flowers, well_memory, n_coin_tossed): #n_coin_tossed correspond aux nénuphares récupérés
  best = float('inf') #On la définie comme infinie pour qu'elle puisse rentrer une première fois dans les tests

  if remaining_flowers == 0: return 0
  #On lance 1 pièces :
  if remaining_flowers >= 1:
    coins = 1 + min_well_coin(remaining_flowers - 1,well_memory,n_coin_tossed + 1)
    if coins < best:
      best = coins
      #print("lance 1")

  #On lance 2 pièces :
  if well_memory > 0 and remaining_flowers >= well_memory : #On regarde si le puit a une "dette en vers nous ET s'il peut honnorer celle-ci"
    coins = 2 + min_well_coin(remaining_flowers - well_memory, well_memory, n_coin_tossed + well_memory)
    if coins < best :
      best = coins
      #print("lance 2")

  #On lance 4 pièces
  if n_coin_tossed != well_memory: #on évite le cas (x,0,0) qui se rappel lui meme a l'infini
    coins = 4 + min_well_coin(remaining_flowers,n_coin_tossed,n_coin_tossed)
    if coins < best:
      best = coins
      #print("lance 4")

  return best
min_well_coin(20, 0, 0)
```

```python
# Les asserts sont flexibles, seul le résultat compte
assert min_well_coin(1, 0, 0) == 1
assert min_well_coin(2, 0, 0) == 2
assert min_well_coin(12, 0, 0) == 12
assert min_well_coin(16, 0, 0) == 14
assert min_well_coin(18, 0, 0) == 14
assert min_well_coin(20, 0, 0) == 15
```

### Explications

  
On test toutes les options de lancer possibles que le puit nous offre selon son nombre de nénuphares ou la "dette" (well_memory) qu'il a en vers nous.

En suite on met à jour le nombre de pièce (best) par rapport au meilleur lancé effectué. (On prend pas en compte les cas bêtes comme lancer 2 pièces si le puis n'a pas de dettes car on veut le lancé de pièces optimal) et on réhitère le proccédé jusqu'à vider le puit.
