## Exercice 1
Concevez un algorithme pour **énumérer tous les arbres binaires** d'une taille N donnée, où N correspond au nombre de noeuds (vertex) ; le N ne prend pas en compte les feuilles. Les noeuds feuilles sont représentés par ceci "(..)": une parenthèse ouverte '(', deux points '..' - représentant les deux feuilles- et une parenthèse fermée ')'.

Le programme final doit afficher la liste des arbres binaires de la taille spécifiée, un arbre par ligne. Pour N=3, il y a un total de 5 arbres binaires. L'output doit être une séquence de parenthèses et points, comme ceci:

    (((..).).)
    ((.(..)).)
    ((..)(..))
    (.((..).))
    (.(.(..)))
    
Implémentez et testez un algorithme récursif qui résout le problème. Donnez une solution pour N=[1,10]

Pour tester votre solution ("est-ce que la solution retourne le nombre correct d'arbres?"), vous devez lire, ~comprendre, implémenter et tester la formule qui calcule le nombre de Catalan pour un N donné:

C<sub>n</sub> = $\frac{(2n)!}{(n+1)!n!}$

Aide: https://fr.wikipedia.org/wiki/Nombre_de_Catalan

```python
import math

def catalan_nb(n: int):
    if n < 0:
      print("Nombre négatif")
      return None
    if n == 0:
      return 1
    try:
      numerateur = math.factorial(2*n)
      denominateur = math.factorial(n+1) * math.factorial(n)
      return numerateur//denominateur
    except:
      print("Erreur")
      return None
```

```python
assert catalan_nb(3) == 5
assert catalan_nb(5) == 42
assert catalan_nb(8) == 1430
```

```python
def get_all_B_Trees(n: int):
    if n == 0:
      return ["."]
    res = []
    for i in range(n):
      gauche = get_all_B_Trees(i)
      droite = get_all_B_Trees(n-1-i)

      for g in gauche:
        for d in droite:
          res.append(f"({g}{d})")
    return res

#test liste
print("Test avec N=[1,10]:")
for i in range(1,11):
  print (f"Taille {i}:")
  print(len(get_all_B_Trees(i)), "=", catalan_nb(i))  #On a décidé de ne pas afficher les arbres car les derniers prendraient trop de place
                                                      #On démontre donc simplement en comparant avec la formule
                                                      #(Les arbres peuvent tout de même s'afficher correctement)
#Test pour N=3
print("Test avec N=3:")
e = 1
for i in get_all_B_Trees(3):
  print(f"{e}: {i}")
  e += 1
```

### Explications
Expliquez comment votre algorithme fonctionne

Fonction catalan_nb():

Nous commençons par importer le module math (pour utiliser la factorielle ensuite). Nous créons ensuite la formule catalan_nb qui va, dans le cas où n > 1, effectuer la formule qui calcule le nombre Catalan en fonction de n. Nous traitons aussi le cas de base ou n=0 et dans ce cas, nous retournons 1. Pour gérer les cas spéciaux: si n < 0, on retourne None, et si le calcul de la factorielle échoue, on affiche "erreur" et on retourne également None.

Fonction get_all_B_Trees():

Dans le cas où n = 0, nous retournons ["."]. Nous créons ensuite une liste vide et créons une boucle in range de N pour construire notre arbre en fonction de N. dans notre bouble, nous commençons à diviser notre arbre en 2 sous parties en utilisant la récursivité (gauche = get_all_B_Trees(i) et droite = get_all_B_Trees(n-1-i)). Comme gauche et droite sont désormais 2 liste distinctes (sous arbre), nous utilisons deux noucvelles boucles imbriquées pour combiner la gauche et la droite et insérer le résultat dans notre liste initialement vide.
Une fois la bouble principale effectuée, nous retournons la liste res qui contient désormais le résultat voulu.
